```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load the excel sheet into a dataframe
df = pd.read_excel('labour.xlsx', header=3, nrows=33)
df
```

```python
print(df.columns.tolist())
```

```python
#Remove additonal string values from year columns
df.columns = df.columns.astype(str).str.strip()

#Rename column from Ontario Labour Productivity to Industry
df.rename(columns={'Ontario Labour Productivity': 'Industry'}, inplace=True)
```

```python
# Set a random seed for reproducibility 
np.random.seed(613)

# Filter relevant columns
years = ['2015', '2016', '2017', '2018', '2019','2020', '2021', '2022', '2023', '2024']
df_subset = df[['Industry'] + years].dropna()

# Select the top 5 most productive industries for 2024
top5 = df_subset.sort_values(by='2024', ascending=False).head(5)

# Plot line chart
plt.figure(figsize=(10, 6))
for _, row in top5.iterrows():
    plt.plot(years, row[years], marker='o', label=row['Industry'])

plt.xlabel('Year')
plt.ylabel('Productivity')
plt.title('Top 5 Industries by Productivity in 2024')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Data source: https://data.ontario.ca/dataset/labour-productivity-for-ontario
