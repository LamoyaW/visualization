```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Set a random seed for reproducibility 
np.random.seed(613)

# Generate sample data
x = np.arange(50)
y = np.random.randint(0,100,50)

# define figure and axes size. plot x and y
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

# label the plot 
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
```

```python
# Define custom font dictionaries for titles and axis labels
font1 = {'family': 'sans-serif', 'color': 'blue', 'size':20}
font2 = {'family': 'monospace', 'color': 'green', 'size' :14}

# Create a line plot
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

# Add titles and labels with custom fonts
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
```

```python
# define figure and axes size.
fig, ax = plt.subplots(figsize=(5,3))

# create a plot of x and y using different markers and line styles 
ax.plot(x,y, 
           marker = '*',
           markersize = 12, 
           color = 'indigo',
           linestyle = '--',
           linewidth = 2,
           markeredgecolor = '#fa9359',
           markerfacecolor = '#000000')

ax.grid(axis = 'y',
        color = 'black',
        linewidth = 2)
```

```python
# import libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# copied code from : https://python-graph-gallery.com/270-basic-bubble-plot/
# create data
df = pd.DataFrame({
      'x': np.random.rand(40),
      'y': np.random.rand(40),
      'z': np.random.rand(40),
   })
df.head()
```

```python
# create scatterplot with modified marker 'D', alpha value and  color 'blue. Added color, linestyle and linewidth.
plt.scatter(df['x'], 
            df['y'], 
            s=df['z']*1000, 
            alpha=0.3,
            marker = 'D',
            color = 'blue',
            linestyle = '--',
             linewidth = 2)

plt.show()
```
